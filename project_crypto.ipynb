```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf

sns.set_style("darkgrid")

print("Cryptocurrency Price Analyzer")
print("Libraries imported successfully!")
```

```python
crypto_symbols = {
    "Bitcoin": "BTC-USD",
    "Ethereum": "ETH-USD",
    "Dogecoin": "DOGE-USD",
    "Solana": "SOL-USD",
    "Binance Coin": "BNB-USD"
}

data = {}

for name, symbol in crypto_symbols.items():

    df = yf.download(
        symbol,
        period="1y",
        interval="1d",
        auto_adjust=True,
        progress=False
    )

    if not df.empty:
        data[name] = df

print("Cryptocurrency data downloaded successfully!")
```

```python
for name, df in data.items():

    print("\n", "="*50)
    print(name)
    print("="*50)

    display(df.head())
```

```python
summary = []

for name, df in data.items():

    prices = df["Close"].dropna()

    summary.append({
        "Cryptocurrency": name,
        "Average Price": np.mean(prices),
        "Maximum Price": np.max(prices),
        "Minimum Price": np.min(prices),
        "Current Price": prices.iloc[-1],
        "Standard Deviation": np.std(prices)
    })

summary_df = pd.DataFrame(summary)

display(summary_df.round(2))
```

```python
performance = []

for name, df in data.items():

    prices = df["Close"].dropna()

    first_price = prices.iloc[0]
    last_price = prices.iloc[-1]

    change = ((last_price - first_price) / first_price) * 100

    performance.append({
        "Cryptocurrency": name,
        "Starting Price": first_price,
        "Latest Price": last_price,
        "Percentage Change": change
    })

performance_df = pd.DataFrame(performance)

display(performance_df.round(2))
```

```python
plt.figure(figsize=(14, 7))

for name, df in data.items():

    plt.plot(
        df.index,
        df["Close"],
        label=name
    )

plt.title("Cryptocurrency Price Comparison")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Standardize the 'Percentage Change' column to float values to allow comparisons
if isinstance(performance_df["Percentage Change"].iloc[0], pd.Series):
    performance_df["Percentage Change"] = performance_df["Percentage Change"].apply(lambda x: float(x.iloc[0]) if hasattr(x, 'iloc') else float(x))

best_crypto = performance_df.loc[
    performance_df["Percentage Change"].idxmax()
]

print("Best Performing Cryptocurrency")
print("--------------------------------")
print("Name:", best_crypto["Cryptocurrency"])
print("Percentage Change:",
      round(float(best_crypto["Percentage Change"]), 2), "%")
```

```python
btc = data["Bitcoin"].copy()

btc["7 Day Moving Average"] = (
    btc["Close"].rolling(window=7).mean()
)

btc["30 Day Moving Average"] = (
    btc["Close"].rolling(window=30).mean()
)

plt.figure(figsize=(14, 6))

plt.plot(btc.index, btc["Close"], label="Bitcoin Price")
plt.plot(btc.index, btc["7 Day Moving Average"],
         label="7 Day Average")

plt.plot(btc.index, btc["30 Day Moving Average"],
         label="30 Day Average")

plt.title("Bitcoin Moving Average Analysis")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()
plt.show()
```
